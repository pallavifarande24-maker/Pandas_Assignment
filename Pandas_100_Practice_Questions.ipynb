### Q44. Fill missing values using mean/median.
> 💡 **Hint:** `df["col"].fillna(df["col"].mean())` or `df["col"].fillna(df["col"].median())`.

---

```python
import pandas as pd
import numpy as np

# Create a sample DataFrame with missing values
df = pd.DataFrame({
    "Rooms": [2, np.nan, 3, 2, 4],
    "Price_k": [200, 250, np.nan, 190, 1200]  # Note the extreme outlier: 1200
})

print("Original DataFrame:")
print(df)

# 1. Fill 'Rooms' with its Mean
rooms_mean = df["Rooms"].mean()
df["Rooms"] = df["Rooms"].fillna(rooms_mean)

# 2. Fill 'Price_k' with its Median
# Sorted values: 190, 200, 250, 1200 -> Median is average of 200 and 250 = 225.0
price_median = df["Price_k"].median()
df["Price_k"] = df["Price_k"].fillna(price_median)

print(f"Calculated Mean for Rooms: {rooms_mean}")
print(f"Calculated Median for Price_k: {price_median}\n")
print("DataFrame after Imputation:")
print(df)
```

### Q56. Detect outliers in a column.
> 💡 **Hint:** Use the **IQR method**: compute Q1, Q3, IQR = Q3–Q1. Outliers are values below `Q1 - 1.5*IQR` or above `Q3 + 1.5*IQR`. Alternatively use Z-score: `|z| > 3`.

---

```python
import pandas as pd

# Sample dataset of salaries with clear outliers (200 and 250)
df = pd.DataFrame({"Salary": [45, 50, 52, 48, 47, 55, 200, 51, 250]})

# Step 1: Compute Q1, Q3, and IQR
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)
IQR = Q3 - Q1

# Step 2: Define boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Step 3: Filter for outliers
outliers_iqr = df[(df["Salary"] < lower_bound) | (df["Salary"] > upper_bound)]

print("Outliers detected using IQR:")
print(outliers_iqr)  

```

### Q98. Compare pivot vs groupby.
> 💡 **Hint:** `groupby()` is more flexible and returns a flat structure. `pivot_table()` gives a 2D grid and handles duplicates with aggregation. `pivot()` does **no aggregation** and fails on duplicates.

---

```python
import pandas as pd

# 1. Create a sample dataset of Store Sales
# Note: There are multiple entries for '2026' + 'North' to demonstrate aggregation.
data = {
    'Year': [2026, 2026, 2026, 2026],
    'Region': ['North', 'North', 'South', 'East'],
    'Sales': [3000, 2000, 3500, 4000]
}
df_sales = pd.DataFrame(data)

# 2. Create a clean, unique dataset for the .pivot() example
# Note: Every combination of Date + Ticker is completely unique (no duplicates).
unique_data = {
    'Date': ['2026-01-01', '2026-01-01', '2026-01-02', '2026-01-02'],
    'Ticker': ['AAPL', 'MSFT', 'AAPL', 'MSFT'],
    'Price': [180, 420, 182, 425]
}
df_stocks = pd.DataFrame(unique_data)

# --- METHOD 1: groupby() ---
print("--- 1. GROUPBY OUTPUT (Flat Table) ---")
# Aggregates the duplicate 'North' regions and flattens the index
groupby_result = df_sales.groupby(['Year', 'Region'])['Sales'].sum().reset_index()
print(groupby_result)
print("\n" + "="*40 + "\n")

# --- METHOD 2: pivot_table() ---
print("--- 2. PIVOT_TABLE OUTPUT (2D Summary Grid) ---")
# Aggregates the duplicate 'North' regions and spreads them into columns
pivot_table_result = df_sales.pivot_table(values='Sales', index='Year', columns='Region', aggfunc='sum')
print(pivot_table_result)
print("\n" + "="*40 + "\n")

# --- METHOD 3: pivot() ---
print("--- 3. PIVOT OUTPUT (2D Reshaped Grid - No Aggregation) ---")
# Reshapes clean data. Will crash if run on df_sales due to duplicate 'North' entries.
pivot_result = df_stocks.pivot(index='Date', columns='Ticker', values='Price')
print(pivot_result)
```

### Q100. Create an end-to-end analysis workflow.
> 💡 **Hint:** Follow this pipeline:  
> 1. Load data → 2. Inspect (shape, dtypes, nulls) → 3. Clean (fillna, drop_duplicates) → 4. Feature engineer → 5. GroupBy/pivot analysis → 6. Sort and rank → 7. Export results.

---

```python
import pandas as pd
import numpy as np

# ==========================================
# STEP 1: LOAD DATA
# ==========================================
# Simulated raw, messy transaction dataset
raw_data = {
    'Transaction_ID': [101, 102, 103, 104, 105, 105, 106, 107], # Note the duplicate 105
    'Date': ['2026-01-15', '2026-01-16', '2026-02-01', '2026-02-15', '2026-03-01', '2026-03-01', '2026-03-02', '2026-03-10'],
    'Store_Location': ['North', 'South', 'North', 'West', 'South', 'South', 'North', 'West'],
    'Category': ['Electronics', 'Clothing', 'Electronics', 'Home', 'Clothing', 'Clothing', 'Home', np.nan], # Missing value
    'Revenue': [1200, 450, 1500, 300, 600, 600, 250, 900],
    'Items_Sold': [3, 5, 4, 2, np.nan, np.nan, 1, 6] # Missing values
}
df = pd.DataFrame(raw_data)

print("--- STEP 1: Data Loaded Successfully ---")

# ==========================================
# STEP 2: INSPECT DATA
# ==========================================
print("\n--- STEP 2: Inspecting Data Structure ---")
print(f"Dataset Shape: {df.shape}")
print("\nData Types:")
print(df.dtypes)
print("\nMissing Values Count:")
print(df.isnull().sum())

# ==========================================
# STEP 3: CLEAN DATA
# ==========================================
print("\n--- STEP 3: Cleaning Data ---")

# A. Drop exact duplicate rows (removes the second transaction 105)
df = df.drop_duplicates()

# B. Handle missing values
# For category, fill missing records with a placeholder string
df['Category'] = df['Category'].fillna('Unknown')
# For items sold, fill missing records with the column median
df['Items_Sold'] = df['Items_Sold'].fillna(df['Items_Sold'].median())

# C. Correct Data Types (Convert Date column to datetime format)
df['Date'] = pd.to_datetime(df['Date'])

# ==========================================
# STEP 4: FEATURE ENGINEERING
# ==========================================
print("\n--- STEP 4: Engineering Features ---")

# Extract Month name from the Date column for seasonal analysis
df['Month'] = df['Date'].dt.strftime('%B')

# Calculate Average Price per Item for each transaction row
df['Avg_Item_Value'] = (df['Revenue'] / df['Items_Sold']).round(2)

# ==========================================
# STEP 5: GROUPBY & PIVOT ANALYSIS
# ==========================================
print("\n--- STEP 5: Performing Pivot Analysis ---")

# Build a summary grid: Total Revenue by Location (rows) across Categories (columns)
summary_report = pd.pivot_table(
    df,
    values='Revenue',
    index='Store_Location',
    columns='Category',
    aggfunc='sum',
    fill_value=0,            # Clean up empty intersecting cells
    margins=True,           # Add row and column grand totals
    margins_name='TOTAL'
)

# ==========================================
# STEP 6: SORT AND RANK
# ==========================================
print("\n--- STEP 6: Sorting and Ranking ---")

# Strip out the 'TOTAL' row before sorting to keep it out of the mix, 
# then sort Locations based on their total performance
final_report = summary_report.drop('TOTAL', axis=0).sort_values(by='TOTAL', ascending=False)

print("\n--- FINAL EXECUTIVE REPORT ---")
print(final_report)

# ==========================================
# STEP 7: EXPORT RESULTS
# ==========================================
print("\n--- STEP 7: Exporting Results ---")

# Save your final data asset to storage formats
final_report.to_csv('store_performance_summary.csv')
# final_report.to_excel('store_performance_summary.xlsx')
print("Successfully saved report to 'store_performance_summary.csv'!")
```
